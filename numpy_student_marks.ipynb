# Day 11: NumPy Student Marks Analysis

## Objective
Analyze student marks using NumPy and visualize the results using Matplotlib.

## Tasks
- Store student marks in a NumPy array.
- Calculate mean, median, maximum, minimum, and standard deviation.
- Identify the highest-scoring and lowest-scoring students.
- Visualize marks using a histogram, bar chart, and box plot.

```python
import numpy as np
import matplotlib.pyplot as plt

print("NumPy version:", np.__version__)
print("Libraries imported successfully!")
```

```python
student_names = [
    "Student 1", "Student 2", "Student 3",
    "Student 4", "Student 5", "Student 6",
    "Student 7", "Student 8", "Student 9",
    "Student 10", "Student 11", "Student 12",
    "Student 13", "Student 14", "Student 15"
]

marks = np.array([
    78, 85, 92, 67, 73,
    88, 90, 81, 76, 95,
    69, 84, 91, 72, 87
])

print("Student Marks Analysis")
print("-" * 30)
print("Total students:", len(marks))
print("Marks:", marks)
```

```python
mean_marks = np.mean(marks)
median_marks = np.median(marks)
maximum_marks = np.max(marks)
minimum_marks = np.min(marks)
std_deviation = np.std(marks)

highest_index = np.argmax(marks)
lowest_index = np.argmin(marks)

print("Mean Marks:", round(mean_marks, 2))
print("Median Marks:", median_marks)
print("Maximum Marks:", maximum_marks)
print("Minimum Marks:", minimum_marks)
print("Standard Deviation:", round(std_deviation, 2))

print("\nHighest Scorer:",
      student_names[highest_index], "-", maximum_marks)

print("Lowest Scorer:",
      student_names[lowest_index], "-", minimum_marks)
```

```python
print("Student-wise Marks")
print("-" * 35)

for name, mark in zip(student_names, marks):
    print(f"{name}: {mark}")

print("\nStudents scoring 80 or above:")
for name, mark in zip(student_names, marks):
    if mark >= 80:
        print(f"{name}: {mark}")
```

```python
plt.figure(figsize=(8, 5))

plt.hist(marks, bins=6, edgecolor="black")
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.bar(student_names, marks)
plt.title("Student-wise Marks Comparison")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.ylim(0, 100)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))

plt.boxplot(marks, patch_artist=True)
plt.title("Box Plot of Student Marks")
plt.ylabel("Marks")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
print("=" * 40)
print("FINAL ANALYSIS SUMMARY")
print("=" * 40)

print(f"Total Students       : {len(marks)}")
print(f"Average Marks        : {np.mean(marks):.2f}")
print(f"Median Marks         : {np.median(marks):.2f}")
print(f"Highest Marks        : {np.max(marks)}")
print(f"Lowest Marks         : {np.min(marks)}")
print(f"Standard Deviation   : {np.std(marks):.2f}")
print(f"Students scoring >=80: {np.sum(marks >= 80)}")

print("\nAnalysis completed successfully!")
```
